```
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
pd.set_option('display.width', 200)
from ml.features.nlp import nlp_features
from ml.features.embeddings import embedding_features
bot = [f'Daily digest #{i}: markets up {i%5}% https://n.example/{i}' for i in range(30)]
hum = [f'{w} went to the {x} yesterday and loved it' for w,x in zip('Ana Bo Chen Dee Eli Fay Gus Hal Ivy Jon'.split()*3, 'market beach gym lake park cafe'.split()*5)]
pd.DataFrame({'bot': {**nlp_features(bot)[0], **embedding_features(bot)}, 'human': {**nlp_features(hum)[0], **embedding_features(hum)}})
```
