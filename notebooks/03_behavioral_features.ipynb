# 03 · Behavioral features
Interval statistics, entropy, burstiness, periodicity (phase concentration), circadian concentration. A perfectly scheduled series vs. an irregular one:

```
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np
pd.set_option('display.width', 200)
from ml.features.behavior import behavior_features
ts = pd.Series(pd.date_range('2026-01-01', periods=100, freq='15min', tz='UTC'))
rng = np.random.default_rng(0)
irr = pd.Series(pd.to_datetime(np.sort(rng.integers(1_700_000_000, 1_700_000_000+86400*20, 100)), unit='s', utc=True))
pd.DataFrame({'scheduled': behavior_features(ts), 'irregular': behavior_features(irr)}).loc[['posts_per_day','coefficient_of_variation','periodicity_score','circadian_concentration','hour_entropy','burstiness_coefficient']]
```

Regularity alone is not proof of automation: news desks and schedulers also look regular. The model combines many signals.
